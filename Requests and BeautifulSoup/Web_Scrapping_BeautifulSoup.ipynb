```python
import requests
from bs4 import BeautifulSoup  
```

```python
url = "https://openweathermap.org/"
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
title = soup.title.string
print("Title of the webpage:", title)
```

```python
# --- 1. Wind Gusts ---
html = """ 
<div class="spaced-content detail">
    <span class="label">Wind Gusts</span>
    <span class="value">33 km/h</span>
</div>
"""

soup = BeautifulSoup(html, 'html.parser')
gusts_div = soup.find('div', class_='spaced-content detail')

if gusts_div:
    label = gusts_div.find('span', class_='label').get_text(strip=True)
    value = gusts_div.find('span', class_='value').get_text(strip=True)
    print(f"{label}: {value}")
else:
    print("Wind Gusts not found.")

# --- 2. RealFeel Shade ---
html = """
<div class="spaced-content detail">
    <span class="label">RealFeel Shade™</span>
    <span class="value">32°</span>
</div>
"""

soup = BeautifulSoup(html, 'html.parser')
shade_div = soup.find('div', class_='spaced-content detail')

if shade_div:
    label = shade_div.find('span', class_='label').get_text(strip=True)
    value = shade_div.find('span', class_='value').get_text(strip=True)
    print(f"{label}: {value}")
else:
    print("RealFeel Shade info not found.")

# --- 3. Air Quality ---
html = """
<div class="spaced-content detail">
    <span class="label">Air Quality</span>
    <span class="value" style="color: #FDB80D">Poor</span>
</div>
"""

soup = BeautifulSoup(html, 'html.parser')
air_quality_div = soup.find('div', class_='spaced-content detail')

if air_quality_div:
    AQ = air_quality_div.find('span', class_='label').get_text(strip=True)
    GPA = air_quality_div.find('span', class_='value').get_text(strip=True)
    print(f"{AQ}: {GPA}")
else:
    print("Couldn't find the Air Quality.")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time

# Optional: Replace this with your actual path to chromedriver
service = Service(executable_path='chromedriver')  # or 'path/to/chromedriver'

# Setup WebDriver
options = webdriver.ChromeOptions()
options.add_argument('--headless')  # Run in background
driver = webdriver.Chrome(service=service, options=options)

# Open the website
url = "https://www.accuweather.com/en/pk/karachi/261158/weather-forecast/261158"
driver.get(url)

# Give the page some time to load
time.sleep(5)

# Use your XPath to locate the target element
try:
    air_quality_block = driver.find_element(By.XPATH, "/html/body/div/div[7]/div[1]/div[1]/a[1]/div[2]/div[2]/div[4]")
    label = air_quality_block.find_element(By.CLASS_NAME, "label").text.strip()
    value = air_quality_block.find_element(By.CLASS_NAME, "value").text.strip()
    print(f"{label}: {value}")
except Exception as e:
    print("Error extracting Air Quality:", e)

# Close the browser
driver.quit()
```
